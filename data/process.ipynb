```python
import pandas as pd
```

```python
df = pd.read_excel("./BerkDataRAW.xlsx")
df
```

```python
df.drop(index=df[df['KREDI'] == 0].index, inplace=True)
df
```

```python
allowed = ["KPD 102", "FİZ 176", "CSE 102L"]

df.drop(
    index=df[
        (df['KREDI'] == 1) & (~df['CourseCode'].isin(allowed))
    ].index,
    inplace=True
)
```

```python
drop_cols = ['FacultyName', 'FacultyCode', 'DepartmentCode', 'CourseId', 'AKTS', 'DERS_ID', 'DONEM_TIP']
df.drop(labels=drop_cols, inplace=True, axis=1)
df
```

```python
df["DepartmentName"] = (
    df["DepartmentName"]
    .str.replace(r'\s*\(.*?\)', '', regex=True)
    .str.strip()                                
)

deps = df["DepartmentName"].unique()
deps
```

```python
df["DepartmentName"] = df["DepartmentName"].replace(
    {"Yapay Zeka ve Veri Mühendisliği": "Bilgisayar Mühendisliği"}
)
```

```python
remove_codes = ["ATA 102", "TDB 102", "TDP 100", "TDP 304", "YBD 102", "YBD 104"]
df.drop(
    index=df[
        df["CourseCode"].isin(remove_codes) |
        df["CourseCode"].str.contains("HOBİ", na=False) |
        df["CourseCode"].str.contains("PFE", na=False)
    ].index,
    inplace=True
)
```

```python

for dep in deps:
    codes = df[df["DepartmentName"] == dep]["CourseCode"].unique()
    for code in codes:
        # Filter the rows with the same DepartmentName and CourseCode
        courses = df[(df["DepartmentName"] == dep) & (df["CourseCode"] == code)]
        if len(courses) > 1:
            sum_count = courses["OgrenciSayisi"].sum()
            
            # Keep the first index, drop the rest
            keep_index = courses.index[0]
            drop_indices = courses.index[1:]
            
            df.loc[keep_index, "OgrenciSayisi"] = sum_count
            df.drop(drop_indices, inplace=True)

# Reset index if necessary
df.reset_index(drop=True, inplace=True)
```
