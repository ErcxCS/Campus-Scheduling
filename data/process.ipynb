```python
import pandas as pd
```

```python
df = pd.read_excel("./BerkData.xlsx")
df
```

```python
for dep1 in df["DepartmentName"].unique():
    print(f"====== Department_1: {dep1} =======")
    dep1_courses = df[df["DepartmentName"] == dep1]["CourseCode"]
    dep1_set = set(dep1_courses)
    for dep2 in df["DepartmentName"].unique():
        if dep1 != dep2:
            print("=============")
            print(f"Department_1: {dep1} vs. Department_2: {dep2}")
            dep2_courses = df[df["DepartmentName"] == dep2]["CourseCode"]
            dep2_set = set(dep2_courses)
            print(dep1_set.intersection(dep2_set))





```

```python
deps = df["DepartmentName"].unique()

for dep in deps:
    codes = df[df["DepartmentName"] == dep]["CourseCode"].unique()
    for code in codes:
        # Filter the rows with the same DepartmentName and CourseCode
        courses = df[(df["DepartmentName"] == dep) & (df["CourseCode"] == code)]
        if len(courses) > 1:
            sum_count = courses["OgrenciSayisi"].sum()
            
            # Keep the first index, drop the rest
            keep_index = courses.index[0]
            drop_indices = courses.index[1:]
            
            df.loc[keep_index, "OgrenciSayisi"] = sum_count
            df.drop(drop_indices, inplace=True)

# Reset index if necessary
df.reset_index(drop=True, inplace=True)
```
